```python
# Q9. Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.
```

```python
# Write your code here

def transpose(matrix):
    transposed = []

    for col in range(len(matrix[0])):
        row = []
        for r in range(len(matrix)):
            row.append(matrix[r][col])
        transposed.append(row)

    return transposed

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Original Matrix:")
for row in matrix:
    print(row)

print("\nTransposed Matrix:")
for row in transpose(matrix):
    print(row)
```

```python
# Q10. Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.
```

```python
# Write your code here

def flatten(nested_list):
    flat = []

    for sublist in nested_list:
        for item in sublist:
            flat.append(item)

    return flat

nested = [[1, 2], [3, 4], [5, 6]]

print("Original:", nested)
print("Flattened:", flatten(nested))
```

```python
# Q13. Write a function that takes a list and returns a new list with the elements in reverse order.
#      Print the original and reversed lists.
```

```python
# Write your code here

def reverse_list(lst):
    return lst[::-1]

numbers = [1, 2, 3, 4, 5]

print("Original:", numbers)
print("Reversed:", reverse_list(numbers))
```

```python
# Q14. Write a function that rotates a list by n positions. Print the original and rotated lists.
```

```python
# Write your code here

def rotate_list(lst, n):
    return lst[n:] + lst[:n]

numbers = [1, 2, 3, 4, 5]

print("Original:", numbers)
print("Rotated Version_1:", rotate_list(numbers, 2))
print("Rotated Version_2:", rotate_list(numbers, 3))
```

```python
# Q12. Write a function that takes two lists and returns a new list containing only the elements that are present in both lists.
#     Print the intersected list.
```

```python
# Write your code here

def intersection(list1, list2):
    return [x for x in list1 if x in list2]

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

print("Intersection:", intersection(list1, list2))
```
