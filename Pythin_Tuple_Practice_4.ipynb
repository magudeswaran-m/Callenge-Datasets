```python
#Q3. Print the first three elements, the last three elements, and the elements from index 2 to 5 of the tuple created in Q1?

print("First three elements:", numbers[:3])
print("Last three elements:", numbers[-3:])
print("Elements from index 2 to 5:", numbers[2:6])
```

```python
#Q4. Create a nested tuple representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column?

matrix = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
print("Matrix:")
for row in matrix:
    print(row)
print("Element at second row and third column:", matrix[1][2])
```

```python
#Q5. Concatenate two tuples: (1, 2, 3) and (4, 5, 6). Print the resulting tuple.

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
result = tuple1 + tuple2
print("Concatenated tuple:", result)
```

```python
#Q6. Create a tuple with duplicate elements and count the occurrences of an element. Find the index of the first occurrence of an element in the tuple.

tuple_with_duplicates = (1, 2, 3, 2, 4, 2, 5)

element_to_count = 2
element_to_find = 2

count = tuple_with_duplicates.count(element_to_count)
index = tuple_with_duplicates.index(element_to_find)

print("Tuple with duplicates:", tuple_with_duplicates)
print(f"Occurrences of {element_to_count}:", count)
print(f"Index of first occurrence of {element_to_find}:", index)
```

```python
#Q7. Create a tuple with 5 elements and unpack it into 5 variables. Print the variables.

Data = (10, 20, 30, 40, 50)
a, b, c, d, e = Data

print("Unpacked variables:")
print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)
print("e =", e)
```

```python
#Q8. Convert a list of the first 5 positive integers to a tuple. Print the tuple.

list_of_integers = [1, 2, 3, 4, 5]

tuple_from_list = tuple(list_of_integers)
print("Tuple from list:", tuple_from_list)
```

```python
#Q9. Create a tuple containing 3 tuples, each with 3 elements. Print the tuple of tuples.

input = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
print("Tuple of tuples:", input)
```

```python
#Q10. Create a tuple with the first 5 positive integers. Convert it to a list, append the number 6, and convert it back to a tuple. Print the resulting tuple.

tuple1 = (1, 2, 3, 4, 5)
list1 = list(tuple1)
list1.append(6)
tuple2 = tuple(list1)
print("Result:", tuple2)
```

```python
#Q11. Create a tuple with the characters of a string. Join the tuple elements into a single string. Print the string.

input = ("H", "e", "l", "l", "o")
result = "".join(input)
print("Output string:", result)
```

```python
#Q12. Create a dictionary with tuple keys and integer values. Print the Dictionary.

students = {
    ("S101",): 85,
    ("S102",): 90,
    ("S103",): 78
}

print("Dictionary:", students)
```

```python
#Q13. Create a nested tuple and iterate over the elements, printing each element.

nested_tuple = ((1, 2), (3, 4), (5, 6))
for sub_tuple in nested_tuple:
    for element in sub_tuple:
        print(element)
```

```python
#Q14. Create a tuple with duplicate elements. Convert it to a set to remove duplicates and print the resulting set.

input = (1, 2, 2, 3, 3, 4)
set1 = set(input)
print("Resulting set:", set1)
```

```python
#Q15. Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.

def get_min_max_sum(t):
    return min(t), max(t), sum(t)

sample_tuple = (1, 2, 3, 4, 5)
min_val, max_val, sum_val = get_min_max_sum(sample_tuple)
print(f"Minimum: {min_val}, Maximum: {max_val}, Sum: {sum_val}")
```
