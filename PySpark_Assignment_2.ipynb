```python
from pyspark.sql.functions import col, regexp_replace, sum

# 1. Clean and cast all three columns to numeric types (as its already in Int format )
d2_cleaned = data2.withColumn("Sales", regexp_replace(col("Sales"), ",", "").cast("double")) \
                  .withColumn("Profit", regexp_replace(col("Profit"), ",", "").cast("double")) \
                  .withColumn("OrderQuantity", regexp_replace(col("OrderQuantity"), ",", "").cast("int"))
```

```python
#10)Write a query to calculate total sales, total profit and total order quantity?
data2.agg(sum("Sales").alias("Total Sales"),\
          sum("Profit").alias("Total Profit"),\
          sum("OrderQuantity").alias("Total Order Quantity")).show()
```

```python
#11)Write a query to calculate the total sales amount, total order quantity for each category.
#Display the category, total sales, and total order qty and order by total sales from highest to lowest?
```

```python
data2.groupBy("Category")\
     .agg(sum("Sales").alias("Total Sales"),\
          sum("OrderQuantity").alias("Total Order Quantity")).orderBy(col("Total Sales").desc()).show()
```

```python
#12)Write a query to calculate the total profit amount for each category. Display the category, total profit, and total order qty
# and order by total profit from highest to lowest?
```

```python
data2.groupby("Category")\
     .agg(sum("Sales").alias("Total Sales"),\
          sum("Profit").alias("Total Profit"),\
          sum("OrderQuantity").alias("Total Order Quantity")).orderBy(col("Total Profit").desc()).show()
```

```python
#13)Write a query to fetch the subcategories where total sales are greater than 100000?
```

```python
data2.groupby("SubCategory")\
     .agg(sum("Sales").alias("Total Sales"))\
     .filter(col("Total Sales")>100000)\
     .select("SubCategory","Total Sales").show()
```

```python
#14)Write a query to fetch the products where total profit is greater than 2500 and sort it
#based on profit from highest to lowest??
```

```python
data2.groupby("ProductName")\
     .agg(sum("Profit").alias("Total_Profit"))\
     .filter(col("Total_Profit") > 2500)\
     .select("ProductName","Total_Profit").show()
```

```python
#15) Write a query to get the total sales and total profit for OƯice Supplies Category?
```

```python
data2.groupby("Category")\
     .agg(sum("Sales").alias("Total Sales"),\
          sum("Profit").alias("Total Profit"))\
     .filter(col("Category")=="Office Supplies")\
     .select("Category","Total Sales","Total Profit").show()
```

```python
#16) Write a query to get the total sales and total profit for Furniture category and  Tables,
#Bookcases sub-categories ?
```

```python
data2.groupby("SubCategory")\
     .agg(sum("Sales").alias("Total Sales"),\
          sum("Profit").alias("Total Profit"))\
     .filter(col("SubCategory") == "Bookcases")\
     .select("SubCategory","Total Sales","Total Profit").show()
```

```python
#17)Write a query to get the total sales and total profit for Technology category and the
#Accessories, Copiers, Phones sub-categories ?
```

```python
data2.groupby("Category", "SubCategory")\
     .agg(sum("Sales").alias("Total Sales"),\
          sum("Profit").alias("Total Profit"))\
     .filter((col("Category") == "Technology") & (col("SubCategory").isin("Accessories", "Copiers", "Phones")))\
     .select("Category","SubCategory","Total Sales","Total Profit").show()
```
