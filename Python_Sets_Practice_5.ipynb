```python
#Q3. Create two sets: one with the first 5 positive integers and another with the first 5 even integers.
#  Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

#Write your code here

set1 = {1, 2, 3, 4, 5}
set2 = {2, 4, 6, 8, 10}

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference (set1 - set2):", set1 - set2)
print("Symmetric Difference:", set1 ^ set2)
```

```python
#Q4. Create a new set containing the squares of the first 10 positive integers using a set comprehension. Print the new set.

#Write your code here

squares = {x**2 for x in range(1, 11)}
print(squares)
```

```python
#Q5. Create a new set containing only the even numbers from the set created in Assignment 1 using a set comprehension. Print the new set.

#Write your code here

even_numbers = {x for x in x if x % 2 == 0}
print(even_numbers)
```

```python
# Sets in Python store only unique elements. Any duplicate values are automatically removed when the set is created. 
# The set() function can be used to remove duplicates from lists, tuples, or other iterable collections.
```

```python
#Q6. Create a set with duplicate elements and remove the duplicates using set methods. Print the modified set.

#Write your code here

duplicate_list = [1, 2, 2, 3, 3, 4, 4, 5, 5]

print("List with duplicates:", duplicate_list)

unique_set = set(duplicate_list)

print("Set without duplicates:", unique_set)
```

```python
#Q7. Create two sets: one with the first 5 positive integers and another with the first 3 positive integers. 
# Check if the second set is a subset of the first set and 
# if the first set is a superset of the second set. Print the results.

set1 = {1, 2, 3, 4, 5}
set2 = {1, 2, 3}

# Check if set2 is a subset of set1
is_subset = set2.issubset(set1)

# Check if set1 is a superset of set2
is_superset = set1.issuperset(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Is Set 2 a subset of Set 1 :", is_subset)
print("Is Set 1 a superset of Set 2 :", is_superset)
```

```python
#Q8. Create a frozenset with the first 5 positive integers. Print the frozenset.

#Write your code here

frozenset1 = frozenset([1, 2, 3, 4, 5])
print("Frozenset:", frozenset1)
```

```python
# A frozenset is an immutable version of a Python set. Once created, elements cannot be added, removed, or modified. 
# It is useful when you need a fixed collection of unique items and can also be used as a dictionary key because it is hashable.
```

```python
#Q9. Create a set with the first 5 positive integers. Convert it to a list, append the number 6, 
# and convert it back to a set. Print the resulting set. 

#Write your code here   

set1 = {1, 2, 3, 4, 5}
list1 = list(set1)
list1.append(6)
set2 = set(list1)
print("Resulting set:", set2)
```

```python
#Q10.Create a dictionary with set keys and integer values. Print the dictionary.

#Write your code here   

set_keys = {1, 2, 3}
int_values = [10, 20, 30]
dict1 = dict(zip(set_keys, int_values))
print("Output Dictionary:", dict1)
```

```python
#Q11. Create a set and iterate over the elements, printing each element.

#Write your code here 

set1 = {1, 2, 3, 4, 5}
for element in set1:
    print("Element:", element)
```

```python
#Q12. Create a set and remove elements from it until it is empty. Print the set after each removal.

set1 = {1, 2, 3, 4, 5}
while set1:
    element = set1.pop()
    print("Removed element:", element)
    print("Set after removal:", set1)
```

```python
#Q13. Create two sets and update the first set with the symmetric difference of the two sets. Print the modified first set.

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

set1.symmetric_difference_update(set2)

print("Modified first set:", set1)
```
