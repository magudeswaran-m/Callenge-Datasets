**Question 4:** Write a Python program to find the largest of three numbers.

```python
# Your code here
data1 = float(input("Enter first number: "))
data2 = float(input("Enter second number: "))
data3 = float(input("Enter third number: "))
if (data1 >= data2) and (data1 >= data3):
  largest = data1
  print("The Largest Number", largest)

elif (data2 >= data1) and (data2 >= data3):
  largest = data2
  print("The Largest Number", largest)

else:
  largest = data3
  print("The Largest Number", largest)
```

**Question 5:** Write a Python program to calculate the factorial of a number


```python
# Your code here
num = int(input("Enter a number: "))
factorial = 1
if num < 0:
   print("Factorial is not defined for negative numbers.")
else:
   for i in range(1, num + 1):
    factorial *= i

print("Factorial of", num, "is", factorial)
```

**Question 11:** Write a Python program to perform arithmetic operations: addition, subtraction, multiplication, and division.

**Question 12:** Write a Python program to demonstrate comparison operators: equal to, not equal to, greater than, less than.

```python
# Your code here
# Define two numbers
x = 30
y = 15

# Comparison Operations
print("x is equal to y :", x == y)   # Equal to
print("x is not equal to y :", x != y)   # Not equal to
print("x is greater than y  :", x > y)    # Greater than
print("x is less than y  :", x < y)    # Less than
```

**Question 13:** Write a Python program to demonstrate logical operators: and, or, not.

```python
# Your code here
# Define two conditions
x = 10
y = 20

# Logical AND
print("x > 5 and y > 15 :", x > 5 and y > 15)

# Logical OR
print("x > 15 or y > 15 :", x > 15 or y > 15)

# Logical NOT
print("not(x > 5) :", not(x > 5))
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
# Your code here
import calendar

year = int(input("Enter a year: "))

if calendar.isleap(year):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

**Question 18:** Write a Python program to reverse a string.

```python
# Your code here

# Define the original string
original_string = "Hello, World!"

# Reverse the string using slicing
reversed_string = original_string[::-1]

print("Original:", original_string)
print("Reversed:", reversed_string)
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here

text = input("Enter a string: ")

if text == text[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```
