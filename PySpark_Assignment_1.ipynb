```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("CSV_Reader").getOrCreate()
data = spark.read.csv("/content/sample_data/Elec_Store.csv", header=True, inferSchema = True)
```

```python
#1)Query the list of columns?
print(list(data.columns))
```

```python
#2)Query the top 5 rows?
data.show(5)
```

```python
#3)Query the bottom 5 rows?
data.orderBy("ProductName", asending = False).limit(5).show()
```

```python
#4)How many total number of orders placed?
data.select("OrderID").count()
print(f"The Total No of Orders Placed are {data.select("OrderID").count()}")
```

```python
#5a)What are the unique channels present and how many are there?
data.select("Channel").distinct().show()
print(f"The Unique Channels are {data.select("Channel").distinct().count()}")
```

```python
#6)How many cities are there?
data.select("City").distinct().count()
print(f"The Total No of Cities are {data.select("City").distinct().count()}")
```

```python
#7)Query the unique Countries and count of the countries?
data.select("City").distinct().show()
print(f"The Unique Countries are {data.select("Country").distinct().count()}")
```

```python
#8)Query the unique Regions and count of the regions?
data.select("Region").distinct().show()
print(f"The Unique Regions are {data.select("Region").distinct().count()}")
```

```python
from pyspark.sql.functions import countDistinct, col
#9)Query the Total cities count by Country and order it from highest count to lowest count?
data.groupBy("Country")\
    .agg(countDistinct("City").alias("city_count"))\
    .orderBy(col("city_count").desc()).show()
```

```python
#10)How many products are there? Display count of products
data.select("ProductName").distinct().show(30)
print(f"The Total No of Products are {data.select("ProductName").distinct().count()}")
```

```python
#11)Query the total sales and total profit?
data.agg(sum("Sales").alias("Total_Sales"),sum("Profit").alias("Total_Profit")).show()
```

```python
#12)Query the total sales by region and country?
data.groupby("Region","Country").agg(sum("Sales").alias("Total_Sales")).show()
```

```python
#13)Query the total order qty by promotion name?
data.groupby("PromotionName").agg(sum("OrderQty").alias("Total_Order_Qty")).show()
```

```python
#14) Query the top 7 countries with sales from highest to lowest order?
data.groupby("Country").agg(sum("Sales").alias("Total_Sales")).orderBy(col("Total_Sales").desc()).limit(7).show()
```

```python
#15)How much the total sales happened in Asia region?
data.filter(col("Region") == "Asia").agg(sum("Sales").alias("Total_Sales")).show()
```

```python
#16)How much total sales happened in Asia region and India?
data.filter((col("Region") == "Asia") & (col("Country") == "India"))\
    .agg(sum("Sales").alias("Total_Sales")).show()
```

```python
#17)How much profit got in Europe region and Denmark country?
data.filter(col("region") == "Europe")\
    .filter(col("Country") == "Denmark")\
    .agg(sum("Profit").alias("Total_Profit")).show()
```

```python
#18) Query the total order qty by Manufacturer and Product Name?
data.groupby("Manufacturer","ProductName")\
    .agg(sum("OrderQty").alias("Total_Order_Qty")).show(30)
```

```python
#19)Query the top 7products with orderqty and sales and show it descending orderbased on total orderqty?
data.groupby("ProductName")\
    .agg(sum("Sales").alias("Total_Sales"),sum("OrderQty").alias("Total_Order_Qty"))\
    .sort(desc("Total_Order_Qty")).show(7, False)
```

```python
#20) Query the total order quantity greater than 10000 by city?
data.groupby("City")\
    .agg(sum("OrderQty").alias("Total_Order_Qty"))\
    .filter(col("Total_Order_Qty") > 10000).show()
```

```python
#21)Query the total sales less than 1000 by product, display it in ascending order?
data.groupby("ProductName")\
    .agg(sum("Sales").alias("Total_Sales"))\
    .filter(col("Total_Sales") < 1000)\
    .sort(asc("Total_Sales")).show()
```

```python
#22)Query the total profit greater than 200000 by product, display it in descending order?
data.groupby("ProductName")\
    .agg(sum("Profit").alias("Total_Profit"))\
    .filter(col("Total_Profit") > 200000)\
    .sort(desc("Total_Profit")).show()
```

```python
# 23)How much total sales happened in China and Beijing?
data.filter(col("City") == "Beijing")\
    .filter(col("Country") == "China")\
    .agg(sum("Sales").alias("Total_Sales")).show()
```

```python
#24) How much total sales happened in Asian Holiday Promotion?
data.filter(col("PromotionName") == "Asian Holiday Promotion")\
    .agg(sum("Sales").alias("Total_Sales")).show()
```

```python
#25) How much total profit value by Contoso, Ltd Manufacturer?
data.filter(col("Manufacturer") == "Contoso, Ltd")\
    .agg(sum("Profit").alias("Total_Profit")).show()
```

```python
#26) Query the total sales and total order qty by product category and order it by Orderqty from highest to lowest?
data.groupby("ProductCategory")\
    .agg(sum("Sales").alias("Total_Sales"),sum("OrderQty").alias("Total_Order_Qty"))\
    .sort(desc("Total_Order_Qty")).show()
```

```python
#27) Query the total sales value in between 1057 and 26700 by Region?
data.groupby("Region")\
    .agg(sum("Sales").alias("Total_Sales"))\
    .filter(col("Total_Sales").between(1057,26700)).show()
```

```python
#28)How many orders placed in Pittsfield city?
data.filter(col("City") == "Pittsfield")\
    .agg(sum("OrderQty").alias("Total_Order_Qty")).show()
```
